**<h3 align="center">$\textbf{Approach and numerical method}$</h3>**

#### Given condition:
The leapfrog particle pusher
A particle with charge-to-mass ratio q/m obeys two coupled first-order ODEs:
$$\frac{dx}{dt} =v,\frac{dv}{dt} = \frac{q}{m} (E(x)+v× B).\qquad\qquad(1)$$

Leapfrog stores position at whole time levels tn = nh and velocity at half levels $t_{n+\frac{1}{2}} = (n+\frac{1}{2})h.$
Each step first updates the velocity using the force at the current position (the kick), then
updates the position using the new velocity (the drift):

$$v_{n+\frac{1}{2}} = v_{n−\frac{1}{2}} +h\frac{q}{m}F_{n} , x_{n+1} = x_{n} +hv_{n+\frac{1}{2}}. \qquad\qquad(2)$$

The staggering makes the method second-order accurate with only one force evaluation per step.
Initial conditions give $v$ at $t = 0$, so the velocity must first be moved back half a step to $v_\frac{-1}{2}$.
Electron plasma oscillation
An electron displaced from equilibrium in a uniform, fixed ion background feels an electric field
proportional to its displacement, and oscillates at the plasma frequency ωp. In normalized units
($ω_p = 1$), the motion is

$$
\frac{dx}{dt} = v, \qquad
\frac{dv}{dt} = -\omega_p^2 x, \qquad
\omega_p = 1
$$

$$
x(0) = 0.5, \qquad
v(0) = 0, \qquad
0 \leq t \leq 20
$$

**Exact solution:**

$$
x(t) = 0.5\cos t, \qquad
v(t) = -0.5\sin t
$$

**<h3 align="center">$\textbf{Numerical algorithm}$</h3>**

```python
"Leapfrog function"
def leapfrog(x_0,v_0,h):
    x = x_0
    v = v_0
    h = h
    a = -x
    
    v_n = v + (h*a)
    x_n = x + (h*v_n)

    
    return (x_n,v_n)
```

the order of accuracy decreases from *second-order* to *first-order*.

Now the **Taylor Series** expansion is:

$$v(t) = v(0) + t \left. \frac{dv}{dt} \right|_{t=0} + \frac{t^2}{2!} \left. \frac{d^2v}{dt^2} \right|_{t=0} + HOT$$

$HOT - Higher order terms$

Staggered start: the system equation for velocity is

$$v_{-1/2} = v_0 + h \frac{dv}{dt}$$

If we compare it with the Taylor Series, we can see that:

$$v_{n+1/2} = v_0 + h \frac{dv}{dt} + HOT$$

This means that it is a **second-order** accurate equation.

Unstaggered start: the system equation for velocity is

$$v_{-1/2} = v_0$$

If we compare it with the Taylor Series, we can see that:

$$v_{n+1/2} = v_0 + h \frac{dv}{dt} + HOT$$

This is a **first-order** accurate equation.

## Finding the Order of the Method

 The equations given in the Leapfrog scheme are:

$$
v_{n+1/2} = v_{n-1/2} + h \cdot a(x_n) \quad \text{   (1)}
$$


$$
x_{n+1} = x_n + h \cdot v_{n + 1/2} \quad \text{   (2)}
$$

Rearranging equation (2) for velocity:

$$
v_{n + 1/2} = \frac{x_{n+1} - x_{n}}{h} \quad \text{   (3)}
$$

Now, we have $v_{n + 1/2}$ in terms of $x$,\
and for $v_{n - 1/2}$ we can use $n = 1 - 1/2 $ in equation (3) \
ie, moving backwards by one time step:

$$
v_{n - 1/2} = \frac{x_{n} - x_{n-1}}{h} \quad \text{   (4)}
$$

Substituting equations (3) and (4) into equation (1), and substituting the electron plasma acceleration $a(x_n) = -\omega_p^2 x_n$, we get:

$$
\frac{x_{n+1} - x_{n}}{h} = \frac{x_{n} - x_{n-1}}{h} - h\omega_p^2 x_n
$$

Multiplying the entire equation by $h$ to clear the denominators:

$$
x_{n+1} - x_{n} = x_{n} - x_{n-1} - \omega_p^2 h^2 x_n
$$


Gathering all position variables onto the left-hand side yields the final central-difference formula:

$$
\boxed{x_{n+1} - 2x_n + x_{n-1} = -\omega_p^2 h^2 x_n}
$$


### **Stability Limit**

The recurrence relation is

$$
x_{n+1} - 2x_n + x_{n-1} = -h^2\omega_p^2x_n
$$

Rearranging,

$$
x_{n+1} + (h^2\omega_p^2 - 2)x_n + x_{n-1} = 0
$$

#### Trial Solution

Let

$$
\boxed{x_n = r^n}
$$

where $r$ is the *amplification factor*.

Therefore,

$$
x_{n+1} = r^{n+1}, \qquad x_{n-1} = r^{n-1}
$$

At $n=1$,

$$
x_1 = r, \qquad x_2 = r^2, \qquad x_0 = 1
$$

Substituting $x_n = r^n$ into the relation,

$$
r^{n+1} + (h^2\omega_p^2 - 2)r^n + r^{n-1} = 0
$$

Dividing throughout by $r^{n-1}$,

$$
\boxed{r^2 + (h^2\omega_p^2 - 2)r + 1 = 0}
$$


The discriminant is

$$
D = b^2 - 4ac
$$

Hence,

$$
D = (h^2\omega_p^2 - 2)^2 - 4
$$

For the system to be stable,$D < 0$

$$
(h^2\omega_p^2 - 2)^2 - 4 < 0
$$

$$
(h^2\omega_p^2 - 2)^2 < 4
$$

$$
-2 < h^2\omega_p^2 - 2 < 2
$$

$$
0 < h^2\omega_p^2 < 4
$$

$$
\boxed{0 < h\omega_p < 2}
$$

The leapfrog method is stable for

$$
\boxed{h\omega_p < 2}
$$

- $h$ is the time-step size.
- $\omega_p$ is the angular frequency of the system.




```python
h=np.array([1.9,1.99,2.01,2.1])     
x=0.5
vi=0
Wp=1
time= 30
a= -(Wp)**2*x
plt.figure(figsize=(12,8))
for j in range(0 , len(h)):
    x=0.5
    vi=0
    Wp=1
    a= -(Wp)**2*x
    n=int(time/h[j])
    p=[]
    p.append(0.5)
    q=[]
    q.append(0)
    t=np.arange(0, time , h[j])
    v=vi - (0.5*h[j]*a)
   
    for i in range(0,n):
        s=leapfrog(x,v,h[j])
        p.append(s[0])
        q.append(s[1])
        x=s[0]
        v=s[1]
    ta=np.array([t[0],t[n]])
    pa=np.array([p[0],p[n]]) 

    plt.subplot(2,2,(j+1))
    plt.plot(t,p,"-o",color="k",label="Position")
    plt.plot(ta,pa,"->",color="r",markersize=10)
    ax = plt.gca()
    ax.spines['left'].set_position('zero')
    ax.spines['bottom'].set_position('zero')
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.xaxis.set_label_coords(0.5, -0.01)
    plt.ylim(-3*np.max(p),3*np.max(p))
    plt.legend(loc="lower right")
    plt.title(f"H={h[j]}")
    plt.xlabel("Time")
    plt.ylabel("Position")
    plt.grid()
plt.show()

#print(t1,p1)
#print(t[0],t[n],p[0],p[n])
```

# Summary and conculsions

The *Leapfrog method* was initially implemented to solve the given initial-value problem.\
The step size was then progressively reduced from ($h$) to ($\frac{h}{2}$) to investigate its effect on the numerical solution.\
The results showed that *reducing the step size improves the accuracy of the numerical solution*, bringing it closer to the exact solution.\
The order of error was $2$ with staggered start and reduced to $1$ with unstaggered start.

The stability of the Leapfrog method is governed by the condition

$$ \omega_p h < 2\ $$

where ($\omega_p$) is the angular frequency of the system and ($h$) is the time step size. If this stability condition is violated, the numerical solution may become unstable.
